```python
def clean_price(price_str):
    if isinstance(price_str, (int, float)):
        return float(price_str)
    
    p = str(price_str)

    p = p.replace('$', '').replace('€', '').replace('USD', '').strip()

    if ',' in p and '.' in p:
        if p.rfind(',') > p.rfind('.'):
            p = p.replace('.', '').replace(',', '.')
        else:
            p = p.replace(',', '')
    elif ',' in p:
        p = p.replace(',', '.')

    return float(p)

df['Unit_Price'] = df['Unit_Price'].apply(clean_price)

display(df.head(10))
print(df['Unit_Price'].describe())
```
