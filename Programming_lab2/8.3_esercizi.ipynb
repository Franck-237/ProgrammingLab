### Es 1
Hai a disposizione un file `data.csv` contenente dati mensili di traffico aereo con due colonne:

- `date`: data in formato `YYYY-MM` (mese/anno)
- `passengers`: numero di passeggeri per quel mese


Costruisci un modello di **regressione polinomiale** che approssima l’andamento del numero di passeggeri nel tempo.

1. Carica il dataset.
2. Convertilo in un formato numerico utilizzando una colonna `mese_numerico` che conti i mesi a partire da gennaio 1949.
3. Applica una regressione polinomiale (grado a tua scelta).
4. Calcola l’RMSE tra i valori reali e quelli predetti.
5. Visualizza i dati reali e la curva stimata con Plotly.

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import root_mean_squared_error
from sklearn.model_selection import train_test_split
import plotly.graph_objects as go


df = pd.read_csv('data.csv')
df.head()
#rimozioni valori nulli
df = df.dropna()
 
#colonna mese_numerico 
df["mese_numerico"] = range(len(df))
 
#X = feature, y = target 
X = df["mese_numerico"].values
y = df["passengers"].values
 
#divisione train/test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
 

#regressione polinomiale grado 3
model_poly = make_pipeline(PolynomialFeatures(3), LinearRegression())
model_poly.fit(X_train.reshape(-1, 1), y_train)
y_pred = model_poly.predict(X_test.reshape(-1, 1))

#predizione su tutto X per la curva del grafico
y_pred_all = model_poly.predict(X.reshape(-1, 1))
 
#RMSE
rmse = root_mean_squared_error(y_test, y_pred)
print(f"RMSE: {rmse}")
 
#visualizzazione con Plotly
fig = go.Figure()
 
fig.add_trace(go.Scatter(
    x=df["mese_numerico"],
    y=df["passengers"],
    name="Dati reali",
    mode='lines+markers',
    line=dict(color="blue", width=2)
))
 
#curva stimata dal modello polinomiale
fig.add_trace(go.Scatter(
    x=df["mese_numerico"],
    y=y_pred_all,
    name=f"Regressione polinomiale grado 3 (RMSE: {rmse})",
    mode='lines',
    line=dict(color="red", width=3, dash='dash')
))
 
fig.update_layout(
    title="traffico aereo mensile - regressione polinomiale",
    xaxis_title="mese numerico",
    yaxis_title="passeggeri"
)
 
fig.show()

```
